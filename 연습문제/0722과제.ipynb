```python
# 01_news_scraping.ipynb

import requests
from bs4 import BeautifulSoup

# 다음 경제 뉴스 URL
url = 'https://news.daum.net/economy'

# 1. 요청 보내기
response = requests.get(url)
response.encoding = 'utf-8'  # UTF-8 인코딩 설정

print(url)
print(type(response))
print(response.status_code)

# 2. HTML 파싱
soup = BeautifulSoup(response.text, 'html.parser')

# 3. 뉴스 리스트 가져오기
li_tags = soup.select('ul.list_newsheadline2 li')
print(type(li_tags), len(li_tags))

# 4. 각 뉴스 제목과 링크 출력
for li_tag in li_tags:
    a_tag = li_tag.find('a')
    link = a_tag['href']

    strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
    title = strong_tag.text.strip()

    print(link)
    print(title)
```

```python
import requests
from bs4 import BeautifulSoup

# 섹션 이름과 URL 키워드를 매핑
section_dict = {
    '기후/환경': 'climate',
    '사회': 'society',
    '경제': 'economy',
    '정치': 'politics',
    '국제': 'world',
    '문화': 'culture',
    '생활': 'life',
    'IT/과학': 'tech',
    '인물': 'people'
}

# 뉴스 출력 함수
def print_news(section_name):
    # 섹션명 유효성 체크
    if section_name not in section_dict:
        print(f"[오류] '{section_name}'은(는) 유효하지 않은 섹션입니다.")
        print("가능한 섹션:", list(section_dict.keys()))
        return

    # URL 생성
    section_eng = section_dict[section_name]
    url = f'https://news.daum.net/{section_eng}'
    print(f"\n======> {url} {section_name} 뉴스 <======")

    # 요청
    response = requests.get(url)
    response.encoding = 'utf-8'

    # 파싱
    soup = BeautifulSoup(response.text, 'html.parser')
    li_tags = soup.select('ul.list_newsheadline2 li')

    # 뉴스 출력
    for li_tag in li_tags:
        a_tag = li_tag.find('a')
        if not a_tag:
            continue
        link = a_tag['href']

        strong_tag = li_tag.select_one('div.cont_thumb strong.tit_txt')
        if strong_tag:
            title = strong_tag.text.strip()
            print(link)
            print(title)

# 함수 호출 예시
print_news('경제')
print_news('사회')
```

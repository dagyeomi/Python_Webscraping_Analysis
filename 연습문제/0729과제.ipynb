```python
import requests
import json
import pandas as pd
import os

client_id = 'UnXIIDYID51p2ZYlnY24'
client_secret = '9Z1WEzpQR3'

headers = {
    'X-Naver-Client-Id': client_id,
    'X-Naver-Client-Secret': client_secret
}

def search_books(query):
    url = 'https://openapi.naver.com/v1/search/book.json'
    payload = {
        'query': query,
        'display': 50,
        'sort': 'sim'
    }
    res = requests.get(url, params=payload, headers=headers)
    if res.ok:
        return res.json()['items']
    else:
        print(f"Error {res.status_code}")
        return []
```

```python
def save_books_to_json(query, filename):
    books = search_books(query)
    if not os.path.isdir("data"):
        os.makedirs("data")
    with open(filename, 'w', encoding='utf-8') as f:
        json.dump(books, f, ensure_ascii=False, indent=4)
    print(f"{len(books)}개 책 저장 완료 → {filename}")
```

```python
def load_books_to_dataframe(filename):
    with open(filename, 'r', encoding='utf-8') as f:
        books = json.load(f)
    return pd.DataFrame(books)
```

```python
query = '파이썬'
file_path = 'data/books.json'
save_books_to_json(query, file_path)
df = load_books_to_dataframe(file_path)
df[['title', 'author', 'publisher', 'discount']].head()
```
